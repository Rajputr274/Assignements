## 19. Find the most repeated character in a string

```python
def most_repeated_char(s):
    try:
        if not s:
            raise ValueError("String is empty")
        counts = {}
        for ch in s:
            counts[ch] = counts.get(ch, 0) + 1
        return max(counts, key=counts.get)
    except (TypeError, ValueError) as e:
        print("Error:", e)

print(most_repeated_char("mississippi"))
```

## 20. Convert lowercase to uppercase without using .upper()

```python
def to_uppercase(s):
    try:
        result = ""
        for ch in s:
            if 'a' <= ch <= 'z':
                result += chr(ord(ch) - 32)
            else:
                result += ch
        return result
    except TypeError as e:
        print("Error: input must be a string.", e)

print(to_uppercase("hello World 123"))
```

## 21. Find the greatest element in a list

```python
def find_max(lst):
    try:
        if not lst:
            raise ValueError("List is empty")
        greatest = lst[0]
        for item in lst:
            if item > greatest:
                greatest = item
        return greatest
    except (TypeError, ValueError) as e:
        print("Error:", e)

print(find_max([3, 9, 2, 45, 12]))
```

## 27. Sort a list without using the built-in sort()

```python
def bubble_sort(lst):
    try:
        arr = lst.copy()
        n = len(arr)
        for i in range(n):
            for j in range(0, n - i - 1):
                if arr[j] > arr[j + 1]:
                    arr[j], arr[j + 1] = arr[j + 1], arr[j]
        return arr
    except TypeError as e:
        print("Error: list elements must be comparable.", e)

print(bubble_sort([5, 3, 8, 1, 9, 2]))
```

## 29. Factorial of a given number

```python
def factorial(n):
    try:
        n = int(n)
        if n < 0:
            raise ValueError("Factorial is not defined for negative numbers")
        result = 1
        for i in range(1, n + 1):
            result *= i
        return result
    except (ValueError, TypeError) as e:
        print("Error:", e)

print(factorial(5))
```

## 30. Check if a number is prime or not

```python
def is_prime(n):
    try:
        n = int(n)
        if n < 2:
            return False
        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False
        return True
    except (ValueError, TypeError) as e:
        print("Error:", e)

print(is_prime(17))
print(is_prime(18))
```

## 31. Find all prime numbers between 1 and 100

```python
def primes_up_to_100():
    try:
        primes = []
        for num in range(2, 101):
            is_p = True
            for i in range(2, int(num ** 0.5) + 1):
                if num % i == 0:
                    is_p = False
                    break
            if is_p:
                primes.append(num)
        return primes
    except Exception as e:
        print("Error:", e)

print(primes_up_to_100())
```

## 32. Sum of digits of a number

```python
def sum_of_digits(n):
    try:
        n = abs(int(n))
        total = 0
        while n > 0:
            total += n % 10
            n //= 10
        return total
    except (ValueError, TypeError) as e:
        print("Error:", e)

print(sum_of_digits(12345))
```

## 33. Fibonacci series up to n terms

```python
def fibonacci_series(n):
    try:
        n = int(n)
        if n <= 0:
            raise ValueError("Number of terms must be positive")
        series = [0, 1]
        while len(series) < n:
            series.append(series[-1] + series[-2])
        return series[:n]
    except (ValueError, TypeError) as e:
        print("Error:", e)

print(fibonacci_series(10))
```

## 34. Check if a number is an Armstrong number

```python
def is_armstrong(n):
    try:
        n = int(n)
        digits = str(n)
        power = len(digits)
        total = sum(int(d) ** power for d in digits)
        return total == n
    except (ValueError, TypeError) as e:
        print("Error:", e)

print(is_armstrong(153))
print(is_armstrong(123))
```

## 39. Guess a randomly generated number between 1 and 100

```python
import random

def guess_number_game(max_attempts=5):
    try:
        target = random.randint(1, 100)
        attempts = 0
        while attempts < max_attempts:
            guess = input("Guess a number between 1 and 100: ")
            try:
                guess = int(guess)
            except ValueError:
                print("Please enter a valid integer.")
                continue
            attempts += 1
            if guess < target:
                print("Too low!")
            elif guess > target:
                print("Too high!")
            else:
                print(f"Correct! The number was {target}.")
                return
        print(f"Out of attempts. The number was {target}.")
    except Exception as e:
        print("Error:", e)

# Uncomment to play interactively:
# guess_number_game()
print("Function defined. Call guess_number_game() to play.")
```

## 40b. Recursion: sum of natural numbers up to n

```python
def sum_natural_numbers(n):
    try:
        n = int(n)
        if n < 0:
            raise ValueError("n must be non-negative")
        if n == 0:
            return 0
        return n + sum_natural_numbers(n - 1)
    except (ValueError, TypeError) as e:
        print("Error:", e)

print(sum_natural_numbers(10))
```

## 44. Return a new list containing only unique elements

```python
def unique_elements(lst):
    try:
        seen = set()
        result = []
        for item in lst:
            if item not in seen:
                seen.add(item)
                result.append(item)
        return result
    except TypeError as e:
        print("Error:", e)

print(unique_elements([1, 2, 2, 3, 3, 3, 4]))
```

## 47. Generate a random password of 8 characters (letters + digits)

```python
import random
import string

def generate_password(length=8):
    try:
        length = int(length)
        chars = string.ascii_letters + string.digits
        return "".join(random.choice(chars) for _ in range(length))
    except (ValueError, TypeError) as e:
        print("Error:", e)

print(generate_password())
```

## 48. Count how many numbers in a list are greater than the average

```python
def count_above_average(lst):
    try:
        if not lst:
            raise ValueError("List is empty")
        avg = sum(lst) / len(lst)
        count = sum(1 for n in lst if n > avg)
        return count
    except (ValueError, TypeError, ZeroDivisionError) as e:
        print("Error:", e)

print(count_above_average([10, 20, 30, 40, 50]))
```

## 49. Menu-based simple calculator using while True and functions

```python
def add_op(a, b): return a + b
def sub_op(a, b): return a - b
def mul_op(a, b): return a * b
def div_op(a, b): return a / b

def calculator():
    operations = {"1": ("Add", add_op), "2": ("Subtract", sub_op),
                  "3": ("Multiply", mul_op), "4": ("Divide", div_op)}
    while True:
        print("\nMenu: 1-Add 2-Subtract 3-Multiply 4-Divide 5-Exit")
        choice = input("Choose an option: ")
        if choice == "5":
            print("Exiting calculator.")
            break
        if choice not in operations:
            print("Invalid option, try again.")
            continue
        try:
            a = float(input("Enter first number: "))
            b = float(input("Enter second number: "))
            name, func = operations[choice]
            result = func(a, b)
            print(f"{name} result: {result}")
        except ValueError as e:
            print("Error: please enter valid numbers.", e)
        except ZeroDivisionError as e:
            print("Error: cannot divide by zero.", e)

# Uncomment to run interactively:
# calculator()
print("Function defined. Call calculator() to run the menu-based calculator.")
```

## 50. Find the greatest (longest) substring/word in a sentence

```python
def greatest_substring(sentence):
    try:
        words = sentence.split()
        if not words:
            raise ValueError("Sentence is empty")
        longest = max(words, key=len)
        return longest
    except (AttributeError, ValueError) as e:
        print("Error:", e)

print(greatest_substring("Python is an amazing programming language"))
```
