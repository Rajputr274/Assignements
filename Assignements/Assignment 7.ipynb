```python
# 15. Write a function to count capital and small letters in a string.
def count_letters(s):
    capital = 0
    small = 0
    for char in s:
        if char.isupper():
            capital += 1
        elif char.islower():
            small += 1
    return capital, small
s="This is the Code of Python Langauge."
count_letters(s)
```

```python
# 19. Write a function to find the most repeated character in a string. 
def most_repeated_char():
    str1=input("enter string to find most repeated char : ")
    print(f"string : {str1}")
    char_count = {}
    for char in str1:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    most_repeated = max(char_count, key=char_count.get)
    return f"The most repeated character is '{most_repeated}' with {char_count[most_repeated]} occurrences."

most_repeated_char()
```

```python
# 20. Write a function to convert all lowercase letters to uppercase without using .upper(). 
def uppper():
    str1=input("enter your text : ")
    result=""
    for i in str1:
        if "a"<= i <="z":
            result+=chr(ord(i)-32)
        else:
            result+=i
    return result

print(uppper())
```

```python
# 24. Write a function to remove duplicates from a list. 

def duplicate():

    lis=[4,3,5,6,2,7,7,1,4,6,5]
    new_lis=[]
    for i in lis:
        if i not in new_lis:
            new_lis.append(i)
        else:
            i+=i

    return new_lis

print(duplicate())
```

```python
# 27. Write a function to sort a list without using the built-in sort() function.

def sort_list():
    lis=[4,3,5,6,2,7,7,1,4,6,5]
    new_lis=[]
    while lis:
        min_value = lis[0]
        for i in lis:
            if i < min_value:
                min_value = i
        new_lis.append(min_value)
        lis.remove(min_value)
    return new_lis

print(sort_list())



```
