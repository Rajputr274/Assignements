## Part 1: Classes & Objects (Basics)

**#1 — Defining a class and creating an object**

```python
class Dog:
    pass

d = Dog()
print(type(d))   # <class '__main__.Dog'>
```

**#2 — The `__init__` constructor**

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

d = Dog("Rex", "Labrador")
d1=Dog("jacky","german")
d2=Dog("Jerry","golden Retriver")
print(d2.name, d2.breed)
```

**#3 — Instance methods**

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        return f"{self.name} says woof!"

d = Dog("Rex")
print(d.bark())
```

**#4 — Instance attributes vs class attributes**

```python
class Dog:
    species = "Canis familiaris"   # class attribute, shared by all instances

    def __init__(self, name):
        self.name = name           # instance attribute, unique per object

a, b = Dog("Rex"), Dog("Fido")
print(a.species is b.species)      # True — same object in memory
print(a.name == b.name)            # False
```

**#5 — Modifying attributes**

```python
d = Dog("Rex")
d.name = "Max"
print(d.name)
```

**#6 — `self` explained**

```python
class Counter:
    def __init__(self):
        self.count = 0

    def increment(self):
        self.count += 1   # `self` refers to the specific instance

c1, c2 = Counter(), Counter()
c1.increment()
print(c1.count, c2.count)   # 1 0 — independent state
```

**#7 — Default parameter values in `__init__`**

```python
class Car:
    def __init__(self, make, model, year=2024):
        self.make = make
        self.model = model
        self.year = year
```

**#8 — `__str__` for readable printing**

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __str__(self):
        return f"Point({self.x}, {self.y})"

print(Point(1, 2))   # Point(1, 2)
```

**#9 — `__repr__` for unambiguous representation**

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __repr__(self):
        return f"Point(x={self.x!r}, y={self.y!r})"

p = Point(1, 2)
p               # in REPL, shows Point(x=1, y=2)
```

**#10 — Checking type and identity**

```python
class Dog: pass
d = Dog()
print(isinstance(d, Dog))   # True
print(type(d) == Dog)       # True
```

## Part 2: Encapsulation

**#11 — Public attributes**

```python
class Account:
    def __init__(self, balance):
        self.balance = balance   # freely accessible
```

**#12 — Protected attributes (convention only)**

```python
class Account:
    def __init__(self, balance):
        self._balance = balance   # "internal use" by convention, not enforced
```

**#13 — Private attributes (name mangling)**

```python
class Account:
    def __init__(self, balance):
        self.__balance = balance   # becomes _Account__balance

a = Account(100)
# print(a.__balance)     # AttributeError
print(a._Account__balance)  # 100 — still reachable, but discouraged
```

**#14 — Getter/setter methods (pre-property style)**

```python
class Account:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount < 0:
            raise ValueError("Balance cannot be negative")
        self.__balance = amount
```

**#15 — `@property` decorator**

```python
class Account:
    def __init__(self, balance):
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, amount):
        if amount < 0:
            raise ValueError("Balance cannot be negative")
        self._balance = amount

a = Account(100)
a.balance = 200      # calls the setter
print(a.balance)     # calls the getter
```

**#16 — Read-only property (no setter)**

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return 3.14159 * self.radius ** 2

c = Circle(5)
print(c.area)
# c.area = 10   # AttributeError: can't set attribute
```

**#17 — `@property.deleter`**

```python
class Account:
    def __init__(self, balance):
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.deleter
    def balance(self):
        print("Deleting balance...")
        del self._balance
```

**#18 — Validating input with properties**

```python
class Person:
    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if not (0 <= value <= 150):
            raise ValueError("Invalid age")
        self._age = value

    def __init__(self, age):
        self.age = age   # triggers the setter's validation
```

## Part 3: Inheritance

**#19 — Basic inheritance**

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

class Dog(Animal):
    def speak(self):
        return f"{self.name} says Woof!"

d = Dog("Rex")
print(d.speak())
```

**#20 — Calling the parent constructor with `super()`**

```python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
```

**#21 — Extending a parent method**

```python
class Animal:
    def info(self):
        return f"Animal: {self.name}"

class Dog(Animal):
    def info(self):
        base = super().info()
        return base + " (a dog)"
```

**#22 — Overriding methods**

```python
class Shape:
    def area(self):
        return 0

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2
```

**#23 — Multi-level inheritance**

```python
class Animal:
    def eat(self): return "eating"

class Mammal(Animal):
    def walk(self): return "walking"

class Dog(Mammal):
    def bark(self): return "barking"

d = Dog()
print(d.eat(), d.walk(), d.bark())
```

**#24 — Multiple inheritance**

```python
class Flyer:
    def fly(self): return "flying"

class Swimmer:
    def swim(self): return "swimming"

class Duck(Flyer, Swimmer):
    pass

duck = Duck()
print(duck.fly(), duck.swim())
```

**#25 — Method Resolution Order (MRO)**

```python
class A:
    def greet(self): return "A"

class B(A):
    def greet(self): return "B"

class C(A):
    def greet(self): return "C"

class D(B, C):
    pass

print(D.mro())      # [D, B, C, A, object]
print(D().greet())  # "B" — leftmost parent wins
```

**#26 — `isinstance()` and `issubclass()`**

```python
class Animal: pass
class Dog(Animal): pass

d = Dog()
print(isinstance(d, Animal))     # True
print(issubclass(Dog, Animal))   # True
```

**#27 — Abstract base classes with `abc`**

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14159 * self.r ** 2

# Shape()   # TypeError: Can't instantiate abstract class
```

**#28 — Mixins**

```python
class JSONMixin:
    def to_json(self):
        import json
        return json.dumps(self.__dict__)

class User(JSONMixin):
    def __init__(self, name):
        self.name = name

print(User("Alice").to_json())
```

**#29 — Overriding `__init__` while keeping parent state**

```python
class Base:
    def __init__(self, x):
        self.x = x

class Derived(Base):
    def __init__(self, x, y):
        super().__init__(x)
        self.y = y
```

**#30 — Checking for method overrides**

```python
class Animal:
    def speak(self): return "generic sound"

class Dog(Animal):
    def speak(self): return "Woof"

print(Dog.speak is Animal.speak)   # False — overridden
```

## Part 4: Polymorphism

**#31 — Polymorphism via a common interface**

```python
class Cat:
    def speak(self): return "Meow"

class Dog:
    def speak(self): return "Woof"

for animal in [Cat(), Dog()]:
    print(animal.speak())
```

**#32 — Duck typing**

```python
class Duck:
    def quack(self): return "Quack!"

class Person:
    def quack(self): return "I'm quacking!"

def make_it_quack(thing):
    print(thing.quack())    # works for anything with a .quack() method

make_it_quack(Duck())
make_it_quack(Person())
```

**#33 — Operator overloading with magic methods**

```python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

print(Vector(1, 2) + Vector(3, 4))   # Vector(4, 6)
```

**#34 — Function polymorphism (built-in)**

```python
print(len("hello"))    # works on strings
print(len([1, 2, 3]))  # and lists
print(len({"a": 1}))   # and dicts
```

**#35 — Polymorphism through inheritance + overriding**

```python
class PaymentMethod:
    def pay(self, amount):
        raise NotImplementedError

class CreditCard(PaymentMethod):
    def pay(self, amount):
        return f"Paid ${amount} via credit card"

class PayPal(PaymentMethod):
    def pay(self, amount):
        return f"Paid ${amount} via PayPal"

for method in [CreditCard(), PayPal()]:
    print(method.pay(50))
```

## Part 5: Magic (Dunder) Methods

**#36 — `__eq__` for equality comparison**

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

print(Point(1, 2) == Point(1, 2))   # True
```

**#37 — `__lt__` and ordering**

```python
class Money:
    def __init__(self, amount):
        self.amount = amount
    def __lt__(self, other):
        return self.amount < other.amount

print(Money(5) < Money(10))
```

**#38 — `__len__`**

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs
    def __len__(self):
        return len(self.songs)

print(len(Playlist(["a", "b", "c"])))
```

**#39 — `__getitem__` and `__setitem__`**

```python
class Playlist:
    def __init__(self):
        self.songs = []
    def __getitem__(self, i):
        return self.songs[i]
    def __setitem__(self, i, value):
        self.songs[i] = value

p = Playlist()
p.songs = ["a", "b"]
print(p[0])
p[0] = "z"
```

**#40 — `__iter__` and `__next__` (custom iterator)**

```python
class Countdown:
    def __init__(self, start):
        self.current = start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        self.current -= 1
        return self.current + 1

for n in Countdown(3):
    print(n)   # 3 2 1
```

**#41 — `__call__` (making an object callable)**

```python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor
    def __call__(self, x):
        return x * self.factor

double = Multiplier(2)
print(double(5))   # 10
```

**#42 — `__contains__`**

```python
class Group:
    def __init__(self, members):
        self.members = members
    def __contains__(self, name):
        return name in self.members

g = Group(["Alice", "Bob"])
print("Alice" in g)   # True
```

**#43 — `__enter__` / `__exit__` (context managers)**

```python
class ManagedFile:
    def __init__(self, path):
        self.path = path
    def __enter__(self):
        self.file = open(self.path, "w")
        return self.file
    def __exit__(self, exc_type, exc_val, exc_tb):
        self.file.close()

with ManagedFile("out.txt") as f:
    f.write("hello")
```

**#44 — `__add__`, `__sub__`, `__mul__` together**

```python
class Vector:
    def __init__(self, x, y): self.x, self.y = x, y
    def __add__(self, o): return Vector(self.x+o.x, self.y+o.y)
    def __sub__(self, o): return Vector(self.x-o.x, self.y-o.y)
    def __mul__(self, k): return Vector(self.x*k, self.y*k)
    def __repr__(self): return f"({self.x}, {self.y})"

v = Vector(1, 2) + Vector(3, 4) - Vector(1, 1)
print(v * 2)
```

**#45 — `__hash__` (making objects usable as dict keys)**

```python
class Point:
    def __init__(self, x, y): self.x, self.y = x, y
    def __eq__(self, o): return (self.x, self.y) == (o.x, o.y)
    def __hash__(self): return hash((self.x, self.y))

d = {Point(1, 2): "origin-ish"}
```

**#46 — `__bool__`**

```python
class Wallet:
    def __init__(self, balance): self.balance = balance
    def __bool__(self):
        return self.balance > 0

w = Wallet(0)
print(bool(w))   # False
```

**#47 — `__del__` (destructor, use sparingly)**

```python
class Resource:
    def __del__(self):
        print("Resource cleaned up")
```

**#48 — `__format__`**

```python
class Temperature:
    def __init__(self, c): self.c = c
    def __format__(self, spec):
        if spec == "f":
            return f"{self.c * 9/5 + 32}°F"
        return f"{self.c}°C"

print(f"{Temperature(100):f}")
```

**#49 — Instance method vs class method vs static method**

```python
class Demo:
    def instance_method(self):
        return f"called from {self}"

    @classmethod
    def class_method(cls):
        return f"called from {cls}"

    @staticmethod
    def static_method():
        return "no access to self or cls"
```

**#50 — `@classmethod` as alternative constructor**

```python
class Pizza:
    def __init__(self, toppings):
        self.toppings = toppings

    @classmethod
    def margherita(cls):
        return cls(["mozzarella", "tomato"])

p = Pizza.margherita()
```

**#51 — `@staticmethod` for utility functions**

```python
class MathUtils:
    @staticmethod
    def is_even(n):
        return n % 2 == 0

print(MathUtils.is_even(4))
```

**#52 — Class variables shared across instances**

```python
class Employee:
    count = 0
    def __init__(self, name):
        self.name = name
        Employee.count += 1

Employee("A"); Employee("B")
print(Employee.count)   # 2
```

**#53 — `__new__` vs `__init__`**

```python
class Singleton:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

a, b = Singleton(), Singleton()
print(a is b)   # True
```

**#54 — `__slots__` for memory efficiency**

```python
class Point:
    __slots__ = ("x", "y")   # no __dict__, less memory, no new attrs allowed
    def __init__(self, x, y):
        self.x, self.y = x, y

p = Point(1, 2)
# p.z = 5   # AttributeError
```

## Part 9: Design Patterns in Python OOP

**#66 — Singleton pattern**

```python
class Config:
    _instance = None
    def __new__(cls, *args, **kwargs):
        if not cls._instance:
            cls._instance = super().__new__(cls)
        return cls._instance
```

**#67 — Factory pattern**

```python
class Dog:
    def speak(self): return "Woof"
class Cat:
    def speak(self): return "Meow"

def animal_factory(kind):
    return {"dog": Dog, "cat": Cat}[kind]()

a = animal_factory("dog")
```

**#68 — Observer pattern**

```python
class Subject:
    def __init__(self):
        self._observers = []
    def subscribe(self, obs):
        self._observers.append(obs)
    def notify(self, event):
        for obs in self._observers:
            obs.update(event)

class Observer:
    def update(self, event):
        print(f"Received: {event}")

s = Subject()
s.subscribe(Observer())
s.notify("data changed")
```

**#69 — Strategy pattern**

```python
class Strategy:
    def execute(self, a, b): raise NotImplementedError

class AddStrategy(Strategy):
    def execute(self, a, b): return a + b

class MultiplyStrategy(Strategy):
    def execute(self, a, b): return a * b

class Calculator:
    def __init__(self, strategy: Strategy):
        self.strategy = strategy
    def calculate(self, a, b):
        return self.strategy.execute(a, b)

calc = Calculator(MultiplyStrategy())
print(calc.calculate(3, 4))
```

**#70 — Decorator pattern (object wrapping, not `@decorator` syntax)**

```python
class Coffee:
    def cost(self): return 5

class MilkDecorator:
    def __init__(self, coffee):
        self.coffee = coffee
    def cost(self):
        return self.coffee.cost() + 1

drink = MilkDecorator(Coffee())
print(drink.cost())   # 6
```

**#71 — Builder pattern**

```python
class House:
    def __init__(self):
        self.parts = []
    def add(self, part):
        self.parts.append(part)
        return self

house = House().add("walls").add("roof").add("windows")
```

**#72 — Adapter pattern**

```python
class EuropeanSocket:
    def voltage(self): return 230

class USPlugAdapter:
    def __init__(self, socket):
        self.socket = socket
    def voltage(self):
        return self.socket.voltage() / 2   # adapts to a different interface

adapter = USPlugAdapter(EuropeanSocket())
```

**#73 — State pattern**

```python
class TrafficLight:
    def __init__(self):
        self.state = "red"
    def next(self):
        self.state = {"red": "green", "green": "yellow", "yellow": "red"}[self.state]

t = TrafficLight()
t.next()
print(t.state)   # green
```

## Part 10: Advanced Topics

**#74 — Metaclasses**

```python
class Meta(type):
    def __new__(mcs, name, bases, namespace):
        namespace["greeting"] = "Hello from metaclass"
        return super().__new__(mcs, name, bases, namespace)

class MyClass(metaclass=Meta):
    pass

print(MyClass().greeting)
```

**#75 — Using metaclasses to enforce rules**

```python
class EnforceMethods(type):
    def __new__(mcs, name, bases, namespace):
        if "run" not in namespace and bases:
            raise TypeError(f"{name} must implement run()")
        return super().__new__(mcs, name, bases, namespace)

class Base(metaclass=EnforceMethods):
    def run(self): pass
```

**#76 — Abstract properties**

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @property
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self, side): self.side = side
    @property
    def area(self):
        return self.side ** 2
```

**#77 — Descriptors (the machinery behind `@property`)**

```python
class PositiveNumber:
    def __set_name__(self, owner, name):
        self.name = "_" + name
    def __get__(self, instance, owner):
        return getattr(instance, self.name)
    def __set__(self, instance, value):
        if value < 0:
            raise ValueError("Must be positive")
        setattr(instance, self.name, value)

class Product:
    price = PositiveNumber()
    def __init__(self, price):
        self.price = price
```

**#78 — `__init_subclass__` (hook for subclass creation)**

```python
class Plugin:
    registry = []
    def __init_subclass__(cls, **kwargs):
        super().__init_subclass__(**kwargs)
        Plugin.registry.append(cls)

class PluginA(Plugin): pass
class PluginB(Plugin): pass

print(Plugin.registry)   # [PluginA, PluginB]
```

**#79 — Protocols (structural typing with `typing.Protocol`)**

```python
from typing import Protocol

class Speaker(Protocol):
    def speak(self) -> str: ...

class Dog:
    def speak(self) -> str:
        return "Woof"

def make_speak(s: Speaker):
    print(s.speak())

make_speak(Dog())   # works even without inheriting Speaker
```

**#80 — Generic classes with `typing.Generic`**

```python
from typing import Generic, TypeVar

T = TypeVar("T")

class Box(Generic[T]):
    def __init__(self, item: T):
        self.item = item

int_box = Box[int](5)
```

**#81 — Multiple dispatch style via `functools.singledispatchmethod`**

```python
from functools import singledispatchmethod

class Printer:
    @singledispatchmethod
    def show(self, arg):
        print(f"generic: {arg}")

    @show.register
    def _(self, arg: int):
        print(f"int: {arg}")

Printer().show(5)       # int: 5
Printer().show("hi")    # generic: hi
```

**#82 — Composition-based dependency injection**

```python
class EmailSender:
    def send(self, msg): print(f"Emailing: {msg}")

class SMSSender:
    def send(self, msg): print(f"Texting: {msg}")

class Notifier:
    def __init__(self, sender):
        self.sender = sender    # injected dependency
    def notify(self, msg):
        self.sender.send(msg)

Notifier(EmailSender()).notify("Hello!")
```

**#83 — Custom exceptions using OOP**

```python
class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        super().__init__(f"Cannot withdraw {amount}, balance is {balance}")
        self.balance = balance
        self.amount = amount

class Account:
    def __init__(self, balance): self.balance = balance
    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError(self.balance, amount)
        self.balance -= amount
```

**#84 — Operator overloading for a custom collection**

```python
class Matrix:
    def __init__(self, data): self.data = data
    def __add__(self, other):
        return Matrix([[a+b for a, b in zip(r1, r2)]
                        for r1, r2 in zip(self.data, other.data)])
    def __repr__(self): return str(self.data)

m1 = Matrix([[1,2],[3,4]])
m2 = Matrix([[5,6],[7,8]])
print(m1 + m2)
```

**#85 — Chained method calls (fluent interface)**

```python
class QueryBuilder:
    def __init__(self):
        self.parts = []
    def select(self, cols):
        self.parts.append(f"SELECT {cols}")
        return self
    def frm(self, table):
        self.parts.append(f"FROM {table}")
        return self
    def build(self):
        return " ".join(self.parts)

q = QueryBuilder().select("*").frm("users").build()
```

**#86 — Weak references to avoid memory leaks**

```python
import weakref

class Node:
    def __init__(self, name):
        self.name = name

n = Node("A")
ref = weakref.ref(n)
print(ref())   # <Node object> while n is alive
```

**#87 — `copy` vs `deepcopy` with custom objects**

```python
import copy

class Container:
    def __init__(self, items):
        self.items = items

c1 = Container([1, 2, 3])
c2 = copy.copy(c1)        # shallow: shares the same list
c3 = copy.deepcopy(c1)    # deep: fully independent list
```

```python
class Point:
    def __init__(self, x, y): self.x, self.y = x, y
    def __copy__(self):
        return Point(self.x, self.y)
```

## Part 11: SOLID Principles Illustrated

**#91 — Single Responsibility Principle**

```python
# Bad: one class does two unrelated jobs
class ReportBad:
    def generate(self): ...
    def save_to_disk(self): ...

# Good: split responsibilities
class ReportGenerator:
    def generate(self): ...

class ReportSaver:
    def save(self, report): ...
```

**#92 — Open/Closed Principle**

```python
class Discount:
    def apply(self, price): return price

class StudentDiscount(Discount):
    def apply(self, price): return price * 0.9

# Adding new discount types doesn't require modifying existing code
```

**#93 — Liskov Substitution Principle**

```python
class Bird:
    def move(self): return "flying"

class Sparrow(Bird):
    pass   # a Sparrow can substitute for a Bird without breaking behavior

# Violation example: Penguin can't "fly", so forcing it under Bird.move() breaks LSP
```

**#94 — Interface Segregation Principle**

```python
from abc import ABC, abstractmethod

class Printer(ABC):
    @abstractmethod
    def print_doc(self): pass

class Scanner(ABC):
    @abstractmethod
    def scan_doc(self): pass

# Classes implement only the interfaces they actually need
class SimplePrinter(Printer):
    def print_doc(self): print("Printing...")
```

**#95 — Dependency Inversion Principle**

```python
from abc import ABC, abstractmethod

class Database(ABC):
    @abstractmethod
    def save(self, data): pass

class MySQLDatabase(Database):
    def save(self, data): print(f"Saving {data} to MySQL")

class App:
    def __init__(self, db: Database):   # depends on abstraction, not concrete class
        self.db = db
```

## Part 12: Putting It Together — A Mini Real-World Example

**#96 — A small library system using multiple OOP concepts**

```python
from abc import ABC, abstractmethod
from dataclasses import dataclass, field

@dataclass
class Book:
    title: str
    author: str
    available: bool = True

class LibraryMember(ABC):
    @abstractmethod
    def borrow_limit(self): pass

class Student(LibraryMember):
    def borrow_limit(self): return 3

class Faculty(LibraryMember):
    def borrow_limit(self): return 10

class Library:
    def __init__(self):
        self.books: list[Book] = []

    def add_book(self, book: Book):
        self.books.append(book)

    def borrow(self, title):
        for book in self.books:
            if book.title == title and book.available:
                book.available = False
                return f"Borrowed: {title}"
        return "Not available"

lib = Library()
lib.add_book(Book("Python 101", "J. Doe"))
print(lib.borrow("Python 101"))
```

**#97 — Adding an observer to notify when a book is returned**

```python
class Library:
    def __init__(self):
        self.books = []
        self.observers = []

    def subscribe(self, fn):
        self.observers.append(fn)

    def return_book(self, title):
        for book in self.books:
            if book.title == title:
                book.available = True
                for fn in self.observers:
                    fn(title)
```

**#98 — Testing polymorphic behavior**

```python
members = [Student(), Faculty()]
for m in members:
    print(m.borrow_limit())
```

**#99 — Combining composition + strategy for pricing**

```python
class PricingStrategy(ABC):
    @abstractmethod
    def price(self, base): pass

class MemberPricing(PricingStrategy):
    def price(self, base): return base * 0.8

class Checkout:
    def __init__(self, strategy: PricingStrategy):
        self.strategy = strategy
    def total(self, base):
        return self.strategy.price(base)

print(Checkout(MemberPricing()).total(100))
```

**#100 — A full class hierarchy summary**

```python
class Vehicle(ABC):
    def __init__(self, make, model):
        self.make, self.model = make, model

    @abstractmethod
    def fuel_type(self): pass

    def __str__(self):
        return f"{self.make} {self.model} ({self.fuel_type()})"


class ElectricCar(Vehicle):
    def fuel_type(self): return "Electric"


class GasCar(Vehicle):
    def fuel_type(self): return "Gasoline"


for v in [ElectricCar("Tesla", "Model 3"), GasCar("Toyota", "Corolla")]:
    print(v)
```

## Quick Reference: Core OOP Concepts

| Concept | What it means |
|---|---|
| **Class** | Blueprint for creating objects |
| **Object** | An instance of a class |
| **Encapsulation** | Bundling data + behavior, restricting direct access |
| **Inheritance** | A class reusing/extending another class's behavior |
| **Polymorphism** | Same interface, different underlying behavior |
| **Abstraction** | Exposing only essential behavior, hiding implementation |
| **Composition** | Building complex objects from simpler ones ("has-a") |
| **MRO** | The order Python searches parent classes for methods |
| **Magic methods** | `__init__`, `__str__`, `__eq__`, etc. — hook into Python's built-in syntax |
| **Metaclass** | The "class of a class" — controls class creation itself |
