## 📌 What is Streamlit?
**Streamlit** is a Python framework that turns plain Python scripts into interactive
web apps — no HTML, CSS, or JavaScript required. It:
- Renders apps via the **rerun model**: every user interaction reruns the script top to bottom
- Provides built-in **widgets** (buttons, sliders, file uploaders, charts, ...)
- Has built-in **layout** primitives (columns, tabs, sidebar, containers)
- Manages app-level state with **`st.session_state`**
- Caches expensive computations with **`@st.cache_data`** / **`@st.cache_resource`**
- Ships native support for **pandas, matplotlib, plotly, and more**

```python
import streamlit as st

st.title("My First App")
name = st.text_input("What's your name?")
if name:
    st.write(f"Hello, {name}!")
```

## ⚠️ How to Run These Examples
Every code cell below is a **complete, standalone Streamlit script**.
Streamlit apps cannot run inside a Jupyter kernel — they must be run from the terminal:

```bash
pip install streamlit
streamlit run app.py
```

To try an example: copy a cell's code into a file (e.g. `app.py`) and run the command above.
Each script is self-contained and syntax-validated.

## 📚 Streamlit API — Quick Reference

### Text & Display
| Function | Description |
|---|---|
| `st.title(text)` | Large page title |
| `st.header(text)` | Section header |
| `st.subheader(text)` | Smaller section header |
| `st.write(obj)` | Universal "magic" display — text, df, chart, dict, etc. |
| `st.markdown(text)` | Render Markdown (and limited HTML) |
| `st.text(text)` | Plain fixed-width text |
| `st.code(code, language)` | Syntax-highlighted code block |
| `st.latex(expr)` | Render LaTeX math |
| `st.caption(text)` | Small gray helper text |
| `st.json(obj)` | Pretty-print JSON |
| `st.metric(label, value, delta)` | KPI-style number with delta arrow |

### Input Widgets
| Function | Description |
|---|---|
| `st.button(label)` | Returns `True` on the rerun right after click |
| `st.checkbox(label)` | Boolean toggle |
| `st.radio(label, options)` | Single choice from a list |
| `st.selectbox(label, options)` | Dropdown single choice |
| `st.multiselect(label, options)` | Dropdown multi choice |
| `st.slider(label, min, max)` | Numeric or range slider |
| `st.select_slider(label, options)` | Slider over discrete labels |
| `st.text_input(label)` | Single-line text |
| `st.text_area(label)` | Multi-line text |
| `st.number_input(label)` | Numeric spinner |
| `st.date_input(label)` | Date picker |
| `st.time_input(label)` | Time picker |
| `st.file_uploader(label)` | Upload a file |
| `st.color_picker(label)` | Pick a hex color |
| `st.toggle(label)` | On/off switch (newer alternative to checkbox) |
| `st.camera_input(label)` | Capture a photo from webcam |
| `st.form(key)` | Group widgets — submits all at once |

### Layout
| Function | Description |
|---|---|
| `st.sidebar` | Put any widget in the left sidebar |
| `st.columns(n)` | Split the page into `n` side-by-side columns |
| `st.tabs(labels)` | Tabbed sections |
| `st.expander(label)` | Collapsible section |
| `st.container()` | Logical grouping/placeholder for content |
| `st.empty()` | Placeholder that can be overwritten/updated |
| `st.divider()` | Horizontal rule |
| `st.popover(label)` | Click-to-reveal popover (newer) |

### Data & Charts
| Function | Description |
|---|---|
| `st.dataframe(df)` | Interactive, sortable table |
| `st.table(df)` | Static table |
| `st.line_chart(df)` | Quick line chart |
| `st.bar_chart(df)` | Quick bar chart |
| `st.area_chart(df)` | Quick area chart |
| `st.scatter_chart(df)` | Quick scatter chart |
| `st.map(df)` | Plot lat/lon points on a map |
| `st.pyplot(fig)` | Render a Matplotlib figure |
| `st.plotly_chart(fig)` | Render a Plotly figure |
| `st.image(data)` | Display an image |
| `st.audio(data)` / `st.video(data)` | Media playback |

### State, Caching & Control Flow
| Function | Description |
|---|---|
| `st.session_state` | Dict-like persistent state across reruns |
| `@st.cache_data` | Cache the *return value* of a function (data) |
| `@st.cache_resource` | Cache a *resource* (DB connection, model, ...) |
| `st.rerun()` | Force an immediate script rerun |
| `st.stop()` | Halt script execution at this point |
| `st.spinner(text)` | Show a spinner while a block runs |
| `st.progress(value)` | Progress bar |
| `st.status(label)` | Multi-step status container (newer) |
| `st.success/info/warning/error(text)` | Colored status boxes |
| `st.balloons()` / `st.snow()` | Celebration animations |
| `st.query_params` | Read/write URL query parameters |
| `st.switch_page(path)` | Navigate to another page (multipage apps) |

````python
# Example 36: session_state — a simple counter that persists across reruns
import streamlit as st

if "count" not in st.session_state:
    st.session_state.count = 0

if st.button("Increment"):
    st.session_state.count += 1

st.write(f"Count: {st.session_state.count}")
````

````python
# Example 37: session_state — initializing multiple values at once
import streamlit as st

defaults = {"score": 0, "lives": 3, "level": 1}
for key, value in defaults.items():
    if key not in st.session_state:
        st.session_state[key] = value

st.write(f"Score: {st.session_state.score} | Lives: {st.session_state.lives} | Level: {st.session_state.level}")
if st.button("Gain a point"):
    st.session_state.score += 10
````

````python
# Example 38: session_state — sharing state between widgets via a callback
import streamlit as st

if "celsius" not in st.session_state:
    st.session_state.celsius = 0.0

def update_fahrenheit():
    st.session_state.fahrenheit = st.session_state.celsius * 9/5 + 32

st.number_input("Celsius", key="celsius", on_change=update_fahrenheit)
st.write(f"Fahrenheit: {st.session_state.get('fahrenheit', 32.0):.1f}")
````

````python
# Example 39: @st.cache_data — cache an expensive computation
import streamlit as st
import time

@st.cache_data
def expensive_computation(n):
    time.sleep(2)  # simulate a slow operation
    return sum(i * i for i in range(n))

st.write("Computing... (first run is slow, reruns are instant)")
result = expensive_computation(1000)
st.write(f"Result: {result}")
````

````python
# Example 40: @st.cache_data — caching a dataframe load function
import streamlit as st
import pandas as pd

@st.cache_data
def load_data():
    # Simulates loading from a file/database
    return pd.DataFrame({
        "Product": ["Widget", "Gadget", "Gizmo"],
        "Price":   [25.0, 150.0, 10.0],
        "Stock":   [100, 30, 200]
    })

df = load_data()
st.dataframe(df)
````

````python
# Example 41: @st.cache_resource — cache a non-data resource (e.g. a connection/model)
import streamlit as st

class FakeModel:
    def __init__(self):
        st.write("Loading model... (this only happens once)")
    def predict(self, x):
        return x * 2

@st.cache_resource
def get_model():
    return FakeModel()

model = get_model()
st.write(f"Prediction for 5: {model.predict(5)}")
````

````python
# Example 60: Building a basic to-do list with session_state
import streamlit as st

if "todos" not in st.session_state:
    st.session_state.todos = []

new_task = st.text_input("New task")
if st.button("Add task") and new_task:
    st.session_state.todos.append(new_task)

for i, task in enumerate(st.session_state.todos):
    col1, col2 = st.columns([4, 1])
    col1.write(f"- {task}")
    if col2.button("Remove", key=f"remove_{i}"):
        st.session_state.todos.pop(i)
        st.rerun()
````

````python
# Example 62: Form validation with custom error messages
import streamlit as st
import re

with st.form("contact_form"):
    email = st.text_input("Email")
    phone = st.text_input("Phone (10 digits)")
    submitted = st.form_submit_button("Submit")

    if submitted:
        errors = []
        if not re.match(r"^[\w.-]+@[\w.-]+\.\w+$", email):
            errors.append("Invalid email format")
        if not re.match(r"^\d{10}$", phone):
            errors.append("Phone must be exactly 10 digits")

        if errors:
            for e in errors:
                st.error(e)
        else:
            st.success("Form submitted successfully!")
````

````python
# Example 90: Implementing undo/redo using a session_state history stack
import streamlit as st

if "history" not in st.session_state:
    st.session_state.history = [""]
    st.session_state.pointer = 0

def current_text():
    return st.session_state.history[st.session_state.pointer]

new_text = st.text_area("Editor", value=current_text())
if st.button("Save change") and new_text != current_text():
    st.session_state.history = st.session_state.history[:st.session_state.pointer+1]
    st.session_state.history.append(new_text)
    st.session_state.pointer += 1

col1, col2 = st.columns(2)
if col1.button("Undo") and st.session_state.pointer > 0:
    st.session_state.pointer -= 1
    st.rerun()
if col2.button("Redo") and st.session_state.pointer < len(st.session_state.history) - 1:
    st.session_state.pointer += 1
    st.rerun()

st.write("Current text:", current_text())
````

````python
# Example 92: Embedding an interactive what-if / sensitivity analysis tool
import streamlit as st
import pandas as pd
import numpy as np

st.title("Loan Payment Calculator")
principal = st.number_input("Loan amount ($)", value=200000)
rate = st.slider("Annual interest rate (%)", 1.0, 15.0, 6.5) / 100 / 12
years = st.slider("Loan term (years)", 5, 30, 20)
n = years * 12

monthly_payment = principal * rate * (1+rate)**n / ((1+rate)**n - 1)
st.metric("Monthly Payment", f"${monthly_payment:,.2f}")

balance = principal
schedule = []
for month in range(1, n+1):
    interest = balance * rate
    principal_paid = monthly_payment - interest
    balance -= principal_paid
    schedule.append({"month": month, "balance": max(balance, 0)})

df = pd.DataFrame(schedule).set_index("month")
st.line_chart(df)
````

````python
# Example 93: Building a simple A/B testing dashboard with statistical summary
import streamlit as st
import numpy as np

st.title("A/B Test Results")
col1, col2 = st.columns(2)
with col1:
    st.subheader("Variant A")
    visitors_a = st.number_input("Visitors A", value=1000)
    conversions_a = st.number_input("Conversions A", value=50)
with col2:
    st.subheader("Variant B")
    visitors_b = st.number_input("Visitors B", value=1000)
    conversions_b = st.number_input("Conversions B", value=65)

rate_a = conversions_a / visitors_a
rate_b = conversions_b / visitors_b
lift = (rate_b - rate_a) / rate_a * 100

st.metric("Conversion Rate A", f"{rate_a:.2%}")
st.metric("Conversion Rate B", f"{rate_b:.2%}", f"{lift:+.1f}%")

if rate_b > rate_a:
    st.success(f"Variant B outperforms A by {lift:.1f}%")
else:
    st.warning("Variant A performs better or equal")
````

### 🌟 Key Concepts to Remember
- Streamlit **reruns the entire script top-to-bottom** every time a widget changes
- Use **`st.session_state`** to persist data across reruns (counters, forms, history)
- **`@st.cache_data`** caches return values (data); **`@st.cache_resource`** caches objects (connections, models) — pick the right one
- **`st.form`** batches multiple widgets so the script only reruns once, on submit
- Always run with `streamlit run app.py` — these scripts will not execute inside a Jupyter cell
- Sidebar (`st.sidebar`), columns (`st.columns`), and tabs (`st.tabs`) are the three main layout tools
