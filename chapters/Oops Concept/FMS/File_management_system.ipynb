# File Management System 
### Main class file_system
### Dedicated Directory - FMS
### 1. Create file
### 2. Update file
### 3. Read all files
### 4. Delete files 


```python
import os

class FileSys:
    def __init__(self):
        

        
        if not os.path.exists("FMS"):
            os.mkdir("FMS")

# 1. Create File -------------------

    def create_file(self):
            file_name = input("Enter file name (without extension): ") + ".txt"
            file_path = os.path.join("FMS", file_name)

            if os.path.exists(file_path):
                print("File already exists.")
                print("*"*50)
                return

            content = input("Enter file content: ")

            with open(file_path, "w+") as file:
                file.write(content)

            print("File created successfully.")

            print("*"*50)


# 2. Upadate file ------------------------------------------------------------------

    def update_file(self):
            file_name = input("Enter file name to update (without extension): ") + ".txt"
            file_path = os.path.join("FMS", file_name)

            if not os.path.exists(file_path):
                print("File does not exist.")
                return
            content = input("Enter content to append: ")
            with open(file_path, "a") as file:
                file.write(content)
                print("Content appended successfully.")

# 3. Read all files --------------------------------------------------------------------

    def read_file(self):
        files=os.listdir("FMS")

        if not files:
            print("file not exists")
            return 

        print("\nAvailable Files:")
        for file in files:
            print(file)

        print("\nFile Contents:\n")

        for file in files:
            file_path = os.path.join("FMS", file)

            print(f"----- {file} -----")

            with open(file_path, "r") as f:
                print(f.read())

            print("*"*50)

# # 4. Delete File -------------------

#     def delete_file(self):
#         file_name = input("Enter file name to delete (without extension): ") + ".txt"
#         file_path = os.path.join("FMS", file_name)

#         if not os.path.exists(file_path):
#             print("File does not exist.")
#             return

#         os.remove(file_path)
#         print("File deleted successfully.")
#         print("----------------------------------------------------------------------")





obj=FileSys()
# obj.create_file()
# obj.update_file()       
obj.read_file()
# obj.delete_file()
```
