Attributes and Methods 
Attributes: Variables that hold data about the object. 
Methods: Functions defined inside a class that describe its behavior.  


```python
class Student: 
    def __init__(self, name, grade): 
        self.name = name  # Attribute 
        self.grade = grade  # Attribute 
    def get_grade(self):  # Method 
        return f"{self.name} is in grade {self.grade}."

# Object creation 
student1 = Student("Rahul",12) 
print(student1.get_grade()) 
```

1. Inheritance  
It is a mechanism in object-oriented programming that allows a class (called a subclass or derived class) to inherit properties and behaviors (attributes and methods) from another class (called a superclass or base class). This promotes code reusability and establishes a hierarchical relationship between classes.

```python
class Instagram_v1:
    messaging = "Available"
    reels = "Available"

    def send_message(self):
        return "Sending Message..."


class Instagram_v2(Instagram_v1):
    video_call = "Available"

    def make_call(self):
        return "Starting Video Call..."


class Meta_Features:
    AI_chat = "Available"


class Instagram_v3(Instagram_v2, Meta_Features):
    notes = "Available"


user = Instagram_v3()

print(user.messaging)
print(user.reels)
print(user.video_call)
print(user.AI_chat)
print(user.notes)
print(user.send_message())
print(user.make_call())
print("created by DheerajAI")
```

```python
class Student:
    def __init__(self):
        self.student = {}

    # to add student to the class -----------------------------
    def add_student(self, name, age, roll_no, city):
        if roll_no in self.student:
            print("Student with this roll number already exists.")
            return
        self.student[roll_no] = {
            "name": name,
            "age": age,
            "city": city
        }
        print(f"Student {name} added successfully.")

    # To get all Students---------------------------------
    
    def get_student(self):
        if not self.student:
            print("Student with this roll number does not exist.")
            return
        for roll_no, details in self.student.items():
            print(f"Roll No: {roll_no}")
            print(f"Name: {details['name']}")
            print(f"Age: {details['age']}")
            print(f"City: {details['city']}")
            print("------------------------")
        

    # To get all student by their Roll No.-----------------------------
    
    def get_student_by_roll_no(self, roll_no):
        if roll_no not in self.student:
            print("Student with this roll number does not exist.")
            return
        details = self.student[roll_no]
        print(f"Roll No: {roll_no}")
        print(f"Name: {details['name']}")
        print(f"Age: {details['age']}")
        print(f"City: {details['city']}")
        print("------------------------")

    
    # To update student by their Roll No.
    
    def update_student(self,roll_no):
        # student_roll=int(input("enter roll no to update : "))
        if roll_no not in self.student:
            print("student with this roll number does not exist.")
            return
        name=input("enter new name : ")
        age=int(input("enter new age : "))
        city=input("enter new city : ")
        self.student[roll_no] = {
            "name": name,
            "age": age,
            "city": city
        }
        print(f"Student {name} updated successfully.")
        print("------------------------")

    # To drop student by serching with Roll No.

    def drop_student(self,roll_no):
        if roll_no not in self.student:
            print("Student with this roll number does not exist.")
            return
        del self.student[roll_no]
        print(f"Student with roll number {roll_no} dropped successfully.")
        print("------------------------")


s= Student()
s.add_student("Dheeraj", 30, 7878, "Gwalior")
s.add_student("Rahul", 30, 9420, "Bamor")
s.add_student("Raghav", 33, 9421, "Nurabaad")
s.add_student("Saumya", 25, 1234, "Nurabaad")
# s.get_student()
# s.get_student_by_roll_no(int(input("enter your roll no : ")))
# s.update_student(int(input("Enter roll number to update student details: ")))
# s.drop_student(int(input("Enter roll number to drop student details: ")))
# s.drop_student(int(input("enter roll no. : ")))
s.get_student()

```

```python
s = Student()

while True:
    print("""
    1. ADD STUDENT
    2. SHOW ALL STUDENTS
    3. SHOW STUDENT BY ROLL NO
    4. UPDATE STUDENT
    5. DELETE STUDENT
    0. EXIT
    """)

    choice = input("Enter your option (1,2,3,4,5,0): ")

    if choice == "0":
        print("Program Closed")
        break

    elif choice == "1":
        print("---------- ADD STUDENT ----------")

        roll_no = int(input("Enter Roll No: "))
        name = input("Enter Name: ")
        age = int(input("Enter Age: "))
        city = input("Enter City: ")

        # Corrected argument order
        s.add_student(name, age, roll_no, city)

    elif choice == "2":
        print("---------- ALL STUDENTS ----------")
        s.get_student()

    elif choice == "3":
        print("------ SEARCH STUDENT ------")

        roll_no = int(input("Enter Roll No: "))
        s.get_student_by_roll_no(roll_no)

    elif choice == "4":
        print("------ UPDATE STUDENT ------")

        roll_no = int(input("Enter Roll No: "))
        s.update_student(roll_no)

    elif choice == "5":
        print("------ DELETE STUDENT ------")

        roll_no = int(input("Enter Roll No: "))
        s.delete_student(roll_no)

    else:
        print("Invalid Choice! Please enter a number between 0 and 5.")
```
