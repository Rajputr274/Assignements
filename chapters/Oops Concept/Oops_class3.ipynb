2.Encapsulation 
--------------------------------------
1. To bind the data members and method into a single unit.
2. Helps in maintainability, readability and usability.
3. Helps maintain data intergity .
 

```python
class Student:
    def __init__(self,name,marks):
        self.name = name # public variable
        self.__marks = marks  # private variable

    def Student_result(self):
        print(f"Name: {self.name}")
        print(f"Marks: {self.__marks}")

# Seetter method to set marks without directly accessing the private variable

    def set_marks(self, marks):
        if 0<= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks. Please enter a value between 0 and 100.")

# Getter method to get marks without directly accessing the private variable
    def get_marks(self):
        return self.__marks
    

s=Student("Rahul", 85)
s.Student_result()

s.set_marks(90)
s.Student_result()

print(s.get_marks())
```

### Name mangling 
It is a technique used in Python to make private variables and methods less accessible from outside the class. In Python, private variables are indicated by prefixing their names with double underscores (e.g., `__marks`). However, they can still be accessed using name mangling, which involves using the class name as a prefix.



```python
print(s._Student__marks) # Accessing private variable using name mangling
```

# 3.Abstraction 
1. It is the process of hiding the internal details of an object and showing only the essential features of the object.
2. It helps in reducing complexity and improving the maintainability of the code.
3. It allows the programmer to focus on what an object does rather than how it does it.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class Credit_Card_Payment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using Credit Card.")

class Upi_Payment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using UPI.")


# pay=Credit_Card_Payment()
pay=Upi_Payment()
pay.pay(1000)
# p=Payment() # This will raise an error because we cannot instantiate an abstract class
# p.pay(1000) 
```

```python
pay=Credit_Card_Payment()
pay.pay(1000)
```

```python

from abc import ABC, abstractmethod
class RBI_Bank(ABC):
    def details(self):
        print("Welcome to RBI Bank")

    def database (self):
        print("Database is Activated for RBI Bank")

    @abstractmethod
    def security(self):
        pass

class SBI(RBI_Bank):
    def details(self):
        super().details()     # for call just above function (connect to above function)
        print("Welcome to SBI Bank")

    def security(self):
        print("SBI Security: OTP+PIN")


class Fraud_Bank(RBI_Bank):
    def details(self):
        super().details()    
        print("Welcome to Fraud Bank")

    def security(self):
        print("Fraud Security: OTP+PIN")

sbi_user=SBI()
sbi_user.details()
sbi_user.database()

f_user=Fraud_Bank()
f_user.details()
```

# 4.Polymorphism

It is the ability of a function, object or method to take many forms. In Python, polymorphism allows us to define methods in the child class with the same name as defined in their parent class.

```python
# 1.Method Overriding

class Payment():
    def pay(self, amount):
        print("generic payment method")

class Upi_payment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using UPI.")


class Card_payment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using Card.")

class Netbanking_payment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using Netbanking.")

for method in [Upi_payment(), Card_payment(), Netbanking_payment()]:
    method.pay(1000)
```

Duck Typing in Python : if walks like a duck and quack like a duck it is duck.


```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")
for animal in [Dog(), Cat()]:
    animal.sound()
```

### 2. Method Overloading 
Its is the same method name with different number of parameters or different type of parameters.
Python does not support method overloading by default. 
But we can achieve it by using default arguments or variable length arguments.

```python
#add(2,3)
# add(2,3,6)

class Calculator:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c

c = Calculator()
c.add(2, 3)  # This will not work as expected because the second add method overrides the first one
```

```python
# using kwargs -----
class Calculator:
    def add(self,*args):
        return sum(args)

c = Calculator()
c.add(2, 3,5) 
```

```python
# Super Method and MRO(Method Resolution Order)---------------

# super () = calls the parent class method form inside child class 
# Used mainly to extend parent behaviour instead of completly overriding




class Person:
    def __init__(self,name,city,age):
        self.name=name
        self.city=city
        self.age=age
class Student(Person):
    def __init__(self,name,roll_no,city,age):
        super().__init__(name,city,age)
        self.name=name
        self.roll_no=roll_no

s=Student("Aman",111,"Gwalior",29)
print(s.name,s.roll_no,s.city,s.age)
            
```
