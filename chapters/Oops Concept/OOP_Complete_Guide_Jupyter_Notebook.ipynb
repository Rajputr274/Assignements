## Chapter 1: Introduction to OOP
Object-Oriented Programming (OOP) organizes code using objects containing attributes and methods.

### Class and Object

```python
class Student:
    pass

s1=Student()
s2=Student()
print(s1)
print(s2)
```

### Constructor

```python
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
s=Student('Rahul',24)
print(s.name,s.age)
```

### Instance/Class/Static Methods

```python
class Student:
    school='ABC'
    def __init__(self,name): self.name=name
    def display(self): print(self.name)
    @classmethod
    def show_school(cls): print(cls.school)
    @staticmethod
    def add(a,b): return a+b

st=Student('Rahul')
st.display();Student.show_school();print(Student.add(2,3))
```

### Encapsulation

```python
class Bank:
    def __init__(self,balance): self.balance=balance
    def deposit(self,a): self.balance+=a
b=Bank(100);b.deposit(50);print(b.balance)
```

### Inheritance

```python
class Father:
    def house(self): print('House')
class Son(Father): pass
Son().house()
```

### Polymorphism

```python
class Dog:
    def sound(self): print('Bark')
class Cat:
    def sound(self): print('Meow')
for a in [Dog(),Cat()]: a.sound()
```

### Abstraction

```python
from abc import ABC,abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def start(self):...
class Car(Vehicle):
    def start(self): print('Car Started')
Car().start()
```

### Dunder Methods

```python
class Student:
    def __init__(self,n): self.n=n
    def __str__(self): return self.n
print(Student('Rahul'))
```

### Operator Overloading

```python
class Number:
    def __init__(self,v): self.v=v
    def __add__(self,o): return self.v+o.v
print(Number(10)+Number(20))
```

### Mini Project

```python
class Student:
    def __init__(self,name,marks): self.name=name; self.marks=marks
    def grade(self): return 'A' if self.marks>=90 else 'B' if self.marks>=75 else 'C' if self.marks>=60 else 'Fail'
    def display(self): print(self.name,self.marks,self.grade())
Student('Rahul',88).display()
```
