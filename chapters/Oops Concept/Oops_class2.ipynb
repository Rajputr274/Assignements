1.Inheritance

 It is a mechanism in which one class acquires the property of another class. Inheritance allows us to define a class that takes all the functionality from a parent class and allows us to add more.

```python
#1.Single Inheritance and Multilevel Inheritance

class Person:
    def show(self):
        print("i am a Person")

class Student(Person):
    def study(self):
        print("i am a Studying")

class TopperStudent(Student): 
    def rank(self):
        print("i got rank 1")


t = TopperStudent()
t.show()  # Output: i got rank 1   
t.study()  # Output: i am a Studying
t.rank()  # Output: i got rank 1
```

```python
# 3.Multiple Inheritance

class Father:
    def skills(self):
        print("Coding")
class Mother:
    def talent(self):
        print("Cooking")
        
class Child(Father, Mother):
    pass

c = Child()
c.skills()  # Output: Coding
c.talent()  # Output: Cooking
```

```python
# 4.Heirarchical Inheritance

class Person:
    def show(self):
        print("i am a Person")

class Student(Person):
    pass

class Teacher(Person):
    pass

class Principal(Person):
    pass


Person().show()
Student().show()
Teacher().show()
Principal().show()
```

```python
# 5. Hybrid Inheritance
# A mix of two or more types of inheritance is called hybrid inheritance. It is a combination of single, multiple, multilevel, and hierarchical inheritance.

class Person:
    def display_person(self):
        print("I am a Person")


# Single Inheritance
class Employee(Person):
    def display_employee(self):
        print("I am an Employee")


# Hierarchical Inheritance
class Manager(Employee):
    def display_manager(self):
        print("I am a Manager")


class Trainer(Employee):
    def display_trainer(self):
        print("I am a Trainer")


# Multiple Inheritance
class TeamLead(Manager, Trainer):
    def display_teamlead(self):
        print("I am a Team Lead")


person = TeamLead()

person.display_person()
person.display_employee()
person.display_manager()
person.display_trainer()
person.display_teamlead()
```
