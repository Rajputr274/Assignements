```python
from abc import ABC, abstractmethod


# Abstract Base Class ----------------------

class Payment(ABC):

    @abstractmethod
    def process_payment(self, amount):
        pass


# Credit Card ----------------------

class CreditCardPayment(Payment):

    def __init__(self, card_number):
        self.card_number = card_number

    def process_payment(self, amount):
        print(f"₹{amount} paid successfully using Credit Card (XXXXXXXXXXXX{self.card_number[-4:]})")


# UPI ----------------------

class UPIPayment(Payment):

    def __init__(self, upi_id):
        self.upi_id = upi_id

    def process_payment(self, amount):
        print(f"₹{amount} paid successfully using UPI ({self.upi_id})")


# Wallet ----------------------

class WalletPayment(Payment):

    def __init__(self, balance):
        self.balance = balance

    def process_payment(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient Wallet Balance!")

        self.balance -= amount

        print(f"₹{amount} paid successfully using Wallet")
        print(f"Remaining Wallet Balance = ₹{self.balance}")


# Checkout ----------------------

class Checkout:

    def checkout(self, payment_method, amount):
        payment_method.process_payment(amount)



# ---------------------- New Payment Method ----------------------

class NetBankingPayment(Payment):

    def __init__(self, bank_name, account_number):
        self.bank_name = bank_name
        self.account_number = account_number

    def process_payment(self, amount):
        print(f"₹{amount} paid successfully using Net Banking")
        print(f"Bank : {self.bank_name}")
        print(f"Account : XXXX{self.account_number[-4:]}")


# ---------------------- Driver Code ----------------------

checkout = Checkout()

print("\n----- Credit Card Payment -----")
credit = CreditCardPayment(input("enter you 16 digit card no : "))
checkout.checkout(credit, 2500)

print("\n----- UPI Payment -----")
upi = UPIPayment("rahul@upi")
checkout.checkout(upi, 1800)

# print("\n----- Wallet Payment -----")
# wallet = WalletPayment(5000)
# checkout.checkout(wallet, 1500)

# print("\n----- Wallet Payment (Insufficient Balance) -----")
# try:
#     checkout.checkout(wallet, 5000)
# except ValueError as e:
#     print("Error:", e)

# print("\n----- Net Banking Payment -----")
# net = NetBankingPayment("State Bank of India", "987654321012")
# checkout.checkout(net, 4000)
```
